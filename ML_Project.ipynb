# **Data Load**

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

#  Data preparation


### Data Preparation as x and y

```python
y = df['logS']
y
```

```python
x = df.drop('logS', axis=1)
x
```

### Data Split


```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100)

x_train
```

```python
x_train
```

```python
x_test
```

# Model Building


## linear Regression

Training the model

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

### Applying the model to make prediction

```python
y_lr_train_pred = lr.predict(x_train)
y_lr_test_pred = lr.predict(x_test)
```

```python
y_lr_test_pred
```

```python
y_lr_train_pred
```

### Evaluate the model preformance

```python
from sklearn.metrics import mean_squared_error, r2_score

lr_train_mea = mean_squared_error(y_train, y_lr_train_pred)
lr_train_r2 = r2_score(y_train, y_lr_train_pred)

lr_test_mea = mean_squared_error(y_test, y_lr_test_pred)
lr_test_r2 = r2_score(y_test, y_lr_test_pred)
```

```python
print('LR MSE (Train): ', lr_train_mea)
print('LR R2 (Train): ', lr_train_r2)
print('LR MSE (TEST): ', lr_test_mea)
print('LR R2 (TEST): ', lr_test_r2)
```

```python
lr_results = pd.DataFrame(['Linear regression', lr_train_mea, lr_train_r2,lr_test_mea,lr_test_r2]).transpose()
lr_results.columns = ['Methods', 'Training MSE', 'Training R2', 'Test MSE','Test R2' ]
```

```python
lr_results
```

## Random Forest

### Training the model

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(max_depth=2, random_state=100)
rf.fit(x_train, y_train)
```

### Aplying the model to make predictions

```python
y_rf_train_pred = lr.predict(x_train)
y_rf_test_pred = lr.predict(x_test)
```

### Evaluate model preformance

```python
from sklearn.metrics import mean_squared_error, r2_score

rf_train_mea = mean_squared_error(y_train, y_rf_train_pred)
rf_train_r2 = r2_score(y_train, y_rf_train_pred)

rf_test_mea = mean_squared_error(y_test, y_rf_test_pred)
rf_test_r2 = r2_score(y_test, y_rf_test_pred)
```

```python
rf_results = pd.DataFrame(['Random Forest', lr_train_mea, lr_train_r2,lr_test_mea,lr_test_r2]).transpose()
rf_results.columns = ['Methods', 'Training MSE', 'Training R2', 'Test MSE','Test R2' ]
rf_results
```

## Model Comparison

```python
df_models = pd.concat([lr_results, rf_results], axis=0)
df_models
```

```python
df_models.reset_index(drop=True)
```
